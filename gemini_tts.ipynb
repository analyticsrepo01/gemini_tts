# Gemini TTS — Latency Benchmarking & Streaming

Demonstrates 4 approaches to TTS generation, focused on achieving **sub-2-second latency** for Viamo:

| Option | Model | Approach | Expected TTFB |
|--------|-------|----------|---------------|
| A | `gemini-2.5-flash-tts` | One-shot baseline | ~30s |
| B | `gemini-2.5-flash-lite-preview-tts` | Chunked → stitch | ~10s |
| C | `gemini-3.1-flash-tts-preview` | Vertex AI streaming | < 2s |
| D | `gemini-3.1-flash-tts-preview` | Cloud TTS `streaming_synthesize` | < 2s |

> **Primary goal**: Options C & D — single continuous stream, downstream system starts playback  
> after first audio chunk without waiting for full generation.  
> Ref: [Gemini TTS docs](https://cloud.google.com/text-to-speech/docs/gemini-tts)

```python
import os
import re
import time
import wave
import queue
import threading
import numpy as np
from pathlib import Path
from IPython.display import Audio, display

from google import genai
from google.genai import types

# ── Vertex AI client (no direct API key needed on GCP)
PROJECT_ID = "my-project-0004-346516"
LOCATION   = "global"

client = genai.Client(vertexai=True, project=PROJECT_ID, location=LOCATION)

OUTPUT_DIR = Path("audio_output")
OUTPUT_DIR.mkdir(exist_ok=True)

SAMPLE_RATE = 24000   # Hz — default for all Gemini TTS models
SAMPLE_WIDTH = 2      # bytes — LINEAR16 / PCM16

print(f"Client ready. Project: {PROJECT_ID}, Location: {LOCATION}")
```

## Helper — Save PCM as WAV

```python
def save_wav(pcm_bytes: bytes, filename: str) -> str:
    path = str(OUTPUT_DIR / filename)
    with wave.open(path, "wb") as wf:
        wf.setnchannels(1)
        wf.setsampwidth(SAMPLE_WIDTH)
        wf.setframerate(SAMPLE_RATE)
        wf.writeframes(pcm_bytes)
    return path

def pcm_duration(pcm_bytes: bytes) -> float:
    return len(pcm_bytes) / (SAMPLE_RATE * SAMPLE_WIDTH)
```

---
## Option A — One-Shot Baseline
**Model**: `gemini-2.5-flash-tts`  
Full audio generated server-side before returning. No streaming. Highest latency.

```python
def option_a_one_shot(text: str, label: str = "test", voice: str = "Aoede") -> dict:
    t0 = time.perf_counter()

    response = client.models.generate_content(
        model="gemini-2.5-flash-tts",
        contents=text,
        config=types.GenerateContentConfig(
            speech_config=types.SpeechConfig(
                voice_config=types.VoiceConfig(
                    prebuilt_voice_config=types.PrebuiltVoiceConfig(voice_name=voice)
                )
            ),
        ),
    )

    total = time.perf_counter() - t0
    pcm = response.candidates[0].content.parts[0].inline_data.data
    path = save_wav(pcm, f"option_a_{label}.wav")

    print(f"[A] TTFB=Total: {total:.2f}s | Audio: {pcm_duration(pcm):.1f}s")
    display(Audio(filename=path, autoplay=False))
    return {"ttfb": total, "total": total, "audio_duration": pcm_duration(pcm)}


results_a = option_a_one_shot(MEDIUM_TEXT, "medium")
```

## Option B — Chunked (Flash Lite)
**Model**: `gemini-2.5-flash-lite-preview-tts`  
Split text at sentence boundaries → sequential API calls → stitch PCM bytes.  
Context window limits to ~20s per chunk; multiple calls required for long audio.

```python
def split_sentences(text: str, max_words: int = 40) -> list:
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())
    chunks, current, word_count = [], [], 0
    for s in sentences:
        w = len(s.split())
        if word_count + w > max_words and current:
            chunks.append(" ".join(current))
            current, word_count = [s], w
        else:
            current.append(s)
            word_count += w
    if current:
        chunks.append(" ".join(current))
    return chunks


def option_b_chunked(text: str, label: str = "test", voice: str = "Aoede") -> dict:
    chunks = split_sentences(text)
    print(f"[B] {len(chunks)} chunks ...")

    t0 = time.perf_counter()
    ttfb = None
    all_pcm = b""

    for i, chunk in enumerate(chunks):
        response = client.models.generate_content(
            model="gemini-2.5-flash-lite-preview-tts",
            contents=chunk,
            config=types.GenerateContentConfig(
                speech_config=types.SpeechConfig(
                    voice_config=types.VoiceConfig(
                        prebuilt_voice_config=types.PrebuiltVoiceConfig(voice_name=voice)
                    )
                ),
            ),
        )
        elapsed = time.perf_counter() - t0
        if ttfb is None:
            ttfb = elapsed
        chunk_pcm = response.candidates[0].content.parts[0].inline_data.data
        all_pcm += chunk_pcm
        print(f"  chunk {i+1}/{len(chunks)}: {pcm_duration(chunk_pcm):.1f}s audio at t={elapsed:.2f}s")

    total = time.perf_counter() - t0
    path = save_wav(all_pcm, f"option_b_{label}.wav")

    print(f"[B] TTFB: {ttfb:.2f}s | Total: {total:.2f}s | Audio: {pcm_duration(all_pcm):.1f}s")
    display(Audio(filename=path, autoplay=False))
    return {"ttfb": ttfb, "total": total, "audio_duration": pcm_duration(all_pcm)}


results_b = option_b_chunked(LONG_TEXT, "long")
```

## Option C — True Streaming via Vertex AI  ⭐
**Model**: `gemini-3.1-flash-tts-preview`  
Single call with `generate_content_stream()` — audio chunks arrive as they are generated.  
Downstream system (Viamo telephony) can begin playback after first chunk without waiting for full audio.

```python
def option_c_streaming(text: str, label: str = "test", voice: str = "Aoede") -> dict:
    t0 = time.perf_counter()
    ttfb = None
    all_pcm = b""
    chunk_count = 0
    buffered_secs = 0.0

    print(f"[C] Streaming '{label}' ...")

    for chunk in client.models.generate_content_stream(
        model="gemini-3.1-flash-tts-preview",
        contents=text,
        config=types.GenerateContentConfig(
            speech_config=types.SpeechConfig(
                voice_config=types.VoiceConfig(
                    prebuilt_voice_config=types.PrebuiltVoiceConfig(voice_name=voice)
                )
            ),
            temperature=1.0,  # 0.0–2.0, Vertex AI only
        ),
    ):
        try:
            pcm = chunk.candidates[0].content.parts[0].inline_data.data
        except (IndexError, AttributeError):
            continue
        if not pcm:
            continue

        now = time.perf_counter() - t0
        if ttfb is None:
            ttfb = now
            print(f"  ✓ First chunk at t={ttfb:.3f}s  ← playback can start here")

        all_pcm += pcm
        chunk_count += 1
        buffered_secs = pcm_duration(all_pcm)

        # Report every 5 seconds of buffered audio
        if chunk_count % 5 == 0:
            print(f"  Buffered {buffered_secs:.1f}s audio at t={now:.2f}s (chunk #{chunk_count})")

    total = time.perf_counter() - t0
    path = save_wav(all_pcm, f"option_c_{label}.wav")

    print(f"[C] TTFB: {ttfb:.3f}s | Total: {total:.2f}s | Audio: {pcm_duration(all_pcm):.1f}s | Chunks: {chunk_count}")
    display(Audio(filename=path, autoplay=False))
    return {"ttfb": ttfb, "total": total, "audio_duration": pcm_duration(all_pcm), "chunks": chunk_count}


results_c = option_c_streaming(LONG_TEXT, "long")
```

## Option C — with Producer/Consumer Threading

Shows how Viamo's telephony layer would consume the stream:  
**Producer** streams chunks into a queue → **Consumer** forwards to downstream WebSocket/telephony without blocking.

```python
def option_c_producer_consumer(text: str, label: str = "test", voice: str = "Aoede") -> dict:
    audio_queue = queue.Queue()
    t0 = time.perf_counter()
    state = {"ttfb": None, "chunks": 0, "all_pcm": b""}

    def producer():
        for chunk in client.models.generate_content_stream(
            model="gemini-3.1-flash-tts-preview",
            contents=text,
            config=types.GenerateContentConfig(
                speech_config=types.SpeechConfig(
                    voice_config=types.VoiceConfig(
                        prebuilt_voice_config=types.PrebuiltVoiceConfig(voice_name=voice)
                    )
                ),
            ),
        ):
            try:
                pcm = chunk.candidates[0].content.parts[0].inline_data.data
            except (IndexError, AttributeError):
                continue
            if pcm:
                if state["ttfb"] is None:
                    state["ttfb"] = time.perf_counter() - t0
                audio_queue.put(pcm)
                state["chunks"] += 1
        audio_queue.put(None)  # sentinel

    def consumer():
        playhead = 0.0
        first = True
        while True:
            pcm = audio_queue.get(timeout=30)
            if pcm is None:
                break
            now = time.perf_counter() - t0
            state["all_pcm"] += pcm
            playhead += pcm_duration(pcm)
            if first:
                print(f"  [consumer] ▶ started playback at t={now:.3f}s → audio position {playhead:.2f}s")
                first = False
            elif state["chunks"] % 5 == 0:
                print(f"  [consumer] playing at t={now:.2f}s → audio position {playhead:.1f}s")

    p = threading.Thread(target=producer, daemon=True)
    c = threading.Thread(target=consumer, daemon=True)
    p.start(); c.start()
    p.join(); c.join()

    total = time.perf_counter() - t0
    path = save_wav(state["all_pcm"], f"option_c_pc_{label}.wav")

    print(f"[C+Consumer] TTFB: {state['ttfb']:.3f}s | Total: {total:.2f}s | Chunks: {state['chunks']}")
    display(Audio(filename=path, autoplay=False))
    return {"ttfb": state["ttfb"], "total": total, "chunks": state["chunks"]}


results_c2 = option_c_producer_consumer(LONG_TEXT, "long")
```

## Option D — Cloud TTS `streaming_synthesize` API
**Model**: `gemini-3.1-flash-tts-preview` via `google-cloud-texttospeech`  
Production streaming: generator yields config + text chunks → server returns audio as it generates.  
Default streaming format: **PCM** (mulaw/alaw also available for telephony).  

> Note: requires Application Default Credentials (`gcloud auth application-default login`) or a service account.

```python
from google.cloud import texttospeech
from google.api_core.client_options import ClientOptions

TTS_LOCATION = "global"
API_ENDPOINT = (
    f"{TTS_LOCATION}-texttospeech.googleapis.com"
    if TTS_LOCATION != "global"
    else "texttospeech.googleapis.com"
)

tts_client = texttospeech.TextToSpeechClient(
    client_options=ClientOptions(api_endpoint=API_ENDPOINT)
)
print(f"Cloud TTS client ready. Endpoint: {API_ENDPOINT}")
```

```python
def option_d_cloud_tts_streaming(
    text: str,
    label: str = "test",
    voice: str = "Aoede",
    model: str = "gemini-3.1-flash-tts-preview",
    prompt: str = "Speak in a calm, professional customer service tone.",
) -> dict:
    """
    Uses Cloud TTS streaming_synthesize — the generator-based API.
    Config is sent once; text chunks stream in as the server produces audio.
    """
    # Split into sentence-level chunks for the generator
    text_chunks = split_sentences(text, max_words=50)
    print(f"[D] {len(text_chunks)} text chunks via streaming_synthesize ...")

    def request_generator():
        # First request: config only
        yield texttospeech.StreamingSynthesizeRequest(
            streaming_config=texttospeech.StreamingSynthesizeConfig(
                voice=texttospeech.VoiceSelectionParams(
                    name=voice,
                    language_code="en-US",
                    model_name=model,
                )
            )
        )
        # Subsequent requests: text chunks (prompt only on first)
        for i, chunk in enumerate(text_chunks):
            yield texttospeech.StreamingSynthesizeRequest(
                input=texttospeech.StreamingSynthesisInput(
                    text=chunk,
                    prompt=prompt if i == 0 else None,
                )
            )

    t0 = time.perf_counter()
    ttfb = None
    all_pcm = b""
    chunk_count = 0

    for response in tts_client.streaming_synthesize(request_generator()):
        now = time.perf_counter() - t0
        pcm = response.audio_content  # raw bytes (PCM by default)
        if pcm:
            if ttfb is None:
                ttfb = now
                print(f"  ✓ First audio chunk at t={ttfb:.3f}s")
            all_pcm += pcm
            chunk_count += 1
            if chunk_count % 5 == 0:
                print(f"  Buffered {pcm_duration(all_pcm):.1f}s audio at t={now:.2f}s")

    total = time.perf_counter() - t0
    path = save_wav(all_pcm, f"option_d_{label}.wav")

    print(f"[D] TTFB: {ttfb:.3f}s | Total: {total:.2f}s | Audio: {pcm_duration(all_pcm):.1f}s | Chunks: {chunk_count}")
    display(Audio(filename=path, autoplay=False))
    return {"ttfb": ttfb, "total": total, "audio_duration": pcm_duration(all_pcm), "chunks": chunk_count}


results_d = option_d_cloud_tts_streaming(LONG_TEXT, "long")
```
